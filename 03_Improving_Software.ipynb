## Exercise 2: Commit Your Current Work

Before using automated tools to modify your code, create a Git commit that captures the current state of your project.

Use a meaningful commit message that describes the work you completed.

As a general rule, avoid:

```bash
git add *
git add .
```

These commands often add files you did not intend to commit.

Instead, be intentional about what you stage. If you want to stage modifications to files Git is already tracking, consider:

```bash
git add -u
```

You can then add any new files individually.

Example:

```bash
git add -u
git add barycentric_triangle.py
git commit -m "Refactored triangle script into reusable library"
```

Avoid vague commit messages such as `update`, `stuff`, `changes`, or `fixes`.

A good commit message should help both you and future collaborators understand what changed and why.
